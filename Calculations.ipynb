```python
def ang_cor(Mono, Source, A = Activity, Ib = Ib):
    Et_511 = Mono[0]/A
    E_511 = Mono[1]/A
    Et_gamma = Mono[2]/A #Et_1275
    E_gamma = Mono[3]/A #E_1275

    WL1 = (1 - Source[0]/(2*A*Ib*E_511*(1-Et_gamma)))/Et_511
    WL2 = (2*Ib*Et_511 + Source[2]/(A*E_gamma) - 1)/((Et_511**2)*Ib)
    Wg = Source[1]/(A*Ib*(E_511**2)*(1-Et_gamma))

    #Double Checking the calculations
    WL1_check = (1 - Source[3]/(2*A*Ib*E_511*E_gamma))/Et_511
    WG_check = Source[4]/(A*Ib*(E_511**2)*E_gamma)

    if(np.abs(WL1 - WL1_check) < 1e-2 and np.abs(Wg - WG_check) < 1e-2):
        print("Calculations are consistent.")
    else:
        print("Calculations are inconsistent.")

    print("WL1: ", WL1, " WL1_check: ", WL1_check)
    print("WL2: ", WL2)
    print("Wg: ", Wg, " WG_check: ", WG_check)
    return np.array([WL1, WL2, Wg]), np.array([WL1_check, WL2, WG_check])
```

```python
def efficiency(W, Mono, Source, A = Activity, Ib = Ib):
    Et_511 = Mono[0]/A
    E_511 = Mono[1]/A
    Et_gamma = Mono[2]/A #Et_1275
    E_gamma = Mono[3]/A #E_1275

    mono_e = np.array([Et_511, E_511, Et_gamma, E_gamma])

    WL1 = W[0]
    WL2 = W[1]
    Wg = W[2]
    print("Using:" , WL1, WL2, Wg)
    
    alpha = Source[0]/Source[1]
    beta = Source[2]/Source[4]
    # print("Alpha: ", alpha)
    # print("Beta: ", beta)
    # print()

    C1 = 4*beta*(WL1**2) - (alpha**2)*WL2*Wg
    C2 = 4*beta*WL1 - (alpha**2)*Wg
    C3 = 4*beta - (alpha**2)*Wg/Ib
    # print("C1: ", C1)
    # print("C2: ", C2)
    # print("C3: ", C3)
    # print()

    S_Et_511 = (C2 - np.sqrt(C2**2 - C1*C3))/C1
    S_E_511 = 2*(1 - WL1*S_Et_511)/(Wg*alpha)
    S_E_1275 = Source[4]/(A*Ib*Wg*(S_E_511**2))
    S_Et_1275 = 1 - Source[1]/(A*Ib*Wg*(S_E_511**2))

    source_e = np.array([S_Et_511, S_E_511, S_Et_1275, S_E_1275])

    #Errors
    Error_Et_511 = np.abs(S_Et_511 - Et_511)/Et_511 * 100
    Error_E_511 = np.abs(S_E_511 - E_511)/E_511 * 100
    Error_Et_1275 = np.abs(S_Et_1275 - Et_gamma)/Et_gamma * 100
    Error_E_1275 = np.abs(S_E_1275 - E_gamma)/E_gamma * 100

    Errors = np.array([Error_Et_511, Error_E_511, Error_Et_1275, Error_E_1275])

    # Create a DataFrame to store the efficiencies
    df = pd.DataFrame({
        'Mono': mono_e,
        'Source': source_e,
        'Error (%)': Errors
    })
    df.index = ['Total Efficiency at 511 keV', 'Photopeak Efficiency at 511 keV', 'Total Efficiency at 1275 keV', 'Photopeak Efficiency at 1275 keV']   
    print(df)

    print("Sum of Errors: ", np.sum(Errors))
    print()

    return df
```
